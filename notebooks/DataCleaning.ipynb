```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)
```

```python
# Remove duplicate rows if any exist
if duplicates > 0:
    df.drop_duplicates(inplace=True)
```

```python
# Convert the Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
# Fill missing values in CouponCode with "No Coupon"
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
# Remove leading and trailing spaces from text columns
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
# Verify that TotalPrice equals Quantity × UnitPrice
df["CalculatedPrice"] = df["Quantity"] * df["UnitPrice"]

# Find rows where the calculated price doesn't match the recorded price
wrong_prices = df[df["CalculatedPrice"] != df["TotalPrice"]]

print("Wrong Prices:", len(wrong_prices))
```

```python
# Check for invalid values
print(df[df["Quantity"] <= 0])
print(df[df["UnitPrice"] <= 0])
print(df[df["TotalPrice"] <= 0])
```

```python
# Remove the temporary validation column
df.drop(columns=["CalculatedPrice"], inplace=True)
```

```python
# Verify that there are no missing values after cleaning
print(df.isnull().sum())
```

```python
# Save the cleaned dataset
df.to_csv("Cleaned_Dataset.csv",index=False)
```
